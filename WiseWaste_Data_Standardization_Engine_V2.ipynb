```
# -*- coding: utf-8 -*-
"""WiseWaste-Data-Standardization-Engine

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/1Lnz1CFSzYIAaXvLZ3U6NOhOcWkZJkhHr

# *WiseWaste-Data-Standardization-Engine*
"""

import io
import os
import re
import zipfile
import numpy as np
import pandas as pd
from google.colab import files

# ============================================================
# 1) CONFIGURATION
# ============================================================

CONFIG = {
    "Inward": {
        "detect_keywords": ["inward"],
        "detect_columns": ["Inward Code", "Received material from", "Project name"],
        "keep_columns": [
            "Inward Code",
            "Date",
            "Received Material From",
            "Vendor Location",
            "Source",
            "Project Name",
            "Vehicle Number",
            "Loading Cost",
            "Driver Name",
            "Vehicle Vendor Name",
            "Transportation Cost",
            "Additional Cost",
            "Total Tax",
            "Material",
            "Received Quantity",
            "Rate",
            "Value of Received Material",
            "Rejected Quantity",
            "Accepted Quantity",
            "Value of Accepted Material",
            "Net Procurement Cost",
            "Payment Status",
            "Weighment Slip",
            "Vendor Invoice",
            "Vehicle GPS Photo",
            "Inward Record Photo",
            "Material GPS Photo",
        ],
        "non_ffill_columns": {
            "Material",
            "Received Quantity",
            "Rate",
            "Amount",
            "Rejected Quantity",
            "Rejected Quantity %",
            "Accepted Billed Qty",
            "Value of Accepted Qty",
        },
        "aliases": {
            "inward code": "Inward Code",
            "date": "Date",
            "received material from": "Received Material From",
            "vendor location": "Vendor Location",
            "source": "Source",
            "project name": "Project Name",
            "vehicle number": "Vehicle Number",
            "loading cost": "Loading Cost",
            "driver name": "Driver Name",
            "vehicle vendor name": "Vehicle Vendor Name",
            "transportation cost": "Transportation Cost",
            "transportation cost in rs": "Transportation Cost",
            "additional cost": "Additional Cost",
            "additional cost in rs": "Additional Cost",
            "total tax": "Total Tax",
            "total tax in": "Total Tax",
            "material": "Material",
            "received quantity": "Received Quantity",
            "received quantity in kg": "Received Quantity",
            "rate": "Rate",
            "rate in rs": "Rate",
            "amount": "Amount",
            "amount in rs": "Amount",
            "rejected qty": "Rejected Quantity",
            "rejected qty in kg": "Rejected Quantity",
            "rejected quantity": "Rejected Quantity",
            "payment status": "Payment Status",
            "weighment slip": "Weighment Slip",
            "vendor invoice": "Vendor Invoice",
            "vehicle gps photo": "Vehicle GPS Photo",
            "inward record photo": "Inward Record Photo",
            "inward record": "Inward Record Photo",
            "material gps photo": "Material GPS Photo",
            "material gps": "Material GPS Photo",

            # extra original columns
            "rejected qty in": "Rejected Quantity %",
            "accepted billed qty in kg": "Accepted Billed Qty",
            "value of accepted qty in rs": "Value of Accepted Qty",
            "total received qty in kg": "Total Received Qty",
            "total accepted qty in kg": "Total Accepted Qty",
            "total rejected qty in kg": "Total Rejected Qty",
            "total amount in rs": "Total Amount",
            "dc number": "DC Number",
            "invoice number": "Invoice Number",
            "s no": "S.no",
            "sno": "S.no",
            "s.no": "S.no",
        },
    },

    "Production": {
        "detect_keywords": ["production"],
        "detect_columns": ["Production Code", "Process-Equipment", "No. of Present Staffs"],
        "keep_columns": [
            "Production Code",
            "Date",
            "Shift",
            "Process-Equipment",
            "Time Taken in Hrs",
            "No. of Staff Present",
            "Material",
            "Quantity in Kg",
            "Total Rejection in Kg",
            "Materials",
            "Material From",
            "Material Quantity",
        ],
        "non_ffill_columns": {
            "Material",
            "Quantity in Kg",
            "Total Rejection in Kg",
            "Materials",
            "Material From",
            "Material Quantity",
        },
        "aliases": {
            "production code": "Production Code",
            "date": "Date",
            "shift": "Shift",
            "process equipment": "Process-Equipment",
            "process-equipment": "Process-Equipment",
            "time taken in hrs": "Time Taken in Hrs",
            "time taken in hrs.": "Time Taken in Hrs",
            "time taken in": "Time Taken in Hrs",
            "time taken in hrs ": "Time Taken in Hrs",
            "time taken in hrs)": "Time Taken in Hrs",
            "time taken (in hrs)": "Time Taken in Hrs",
            "no of present staffs": "No. of Staff Present",
            "no. of present staffs": "No. of Staff Present",
            "no. of staff present": "No. of Staff Present",
            "material": "Material",
            "quantity in kg": "Quantity in Kg",
            "quantity (in kg)": "Quantity in Kg",
            "total rejection in kg": "Total Rejection in Kg",
            "total rejection (in kg)": "Total Rejection in Kg",
            "materials": "Materials",
            "material from": "Material From",
            "material quantity": "Material Quantity",
            "material quantity in kg": "Material Quantity",
            "material quantity (in kg)": "Material Quantity",

            # extra original columns
            "total quantity in kg": "Total Quantity",
            "total quantity (in kg)": "Total Quantity",
            "description": "Description",
            "s no": "S.no",
            "sno": "S.no",
            "s.no": "S.no",
        },
    },

    "Outward": {
        "detect_keywords": ["outward"],
        "detect_columns": ["Outward Code", "Vehicle number", "Dispatched Qty (in KG)"],
        "keep_columns": [
            "Outward Code",
            "Date",
            "Project",
            "Customer",
            "Destination",
            "Vehicle Number",
            "Transport Vendor",
            "Driver Name",
            "Driver Contact No",
            "Loading Cost",
            "Additional Transport Cost",
            "Transportation Cost",
            "Material",
            "Dispatched Quantity",
            "Rate",
            "Value of Dispatched Material",
            "Rejected Quantity",
            "Accepted Quantity",
            "Value of Accepted Material",
            "Tax",
            "Payment Status",
            "Incentive Cost per Kg",
            "Total Incentive Cost",
            "Net Material Sales Cost",
            "GPS Photo 1",
            "GPS Photo 2",
            "Invoice",
            "LR Copy",
            "DC Challan",
            "Weighment Slip",
            "E-Way Bill",
            "Received Acknowledgement Form ED",
            "Vehicle and Material Photo",
        ],
        "non_ffill_columns": {
            "Material",
            "Dispatched Quantity",
            "Rate",
            "Billed Amount",
            "Rejected Quantity",
            "Tax",
            "Incentive Cost per Kg",
        },
        "aliases": {
            "outward code": "Outward Code",
            "date": "Date",
            "project": "Project",
            "customer": "Customer",
            "destination": "Destination",
            "vehicle number": "Vehicle Number",
            "vehicle no": "Vehicle Number",
            "transport vendor": "Transport Vendor",
            "driver name": "Driver Name",
            "driver contact no": "Driver Contact No",
            "loading cost": "Loading Cost",
            "additional transport cost": "Additional Transport Cost",
            "transportation cost": "Transportation Cost",
            "transport cost": "Transportation Cost",
            "transport cost in rs": "Transportation Cost",
            "material": "Material",
            "dispatched quantity": "Dispatched Quantity",
            "dispatched qty in kg": "Dispatched Quantity",
            "dispatched qty": "Dispatched Quantity",
            "rate": "Rate",
            "rate in rs": "Rate",
            "billed amount": "Billed Amount",
            "billed amount in rs": "Billed Amount",
            "rejected quantity": "Rejected Quantity",
            "rejection qty in kg": "Rejected Quantity",
            "rejection qty": "Rejected Quantity",
            "tax": "Tax",
            "total tax in": "Tax",
            "payment status": "Payment Status",
            "incentive cost per kg": "Incentive Cost per Kg",
            "incentive costkg": "Incentive Cost per Kg",
            "gps photo 1": "GPS Photo 1",
            "gsp photo 1": "GPS Photo 1",
            "gps photo 2": "GPS Photo 2",
            "gsp photo 2": "GPS Photo 2",
            "invoice": "Invoice",
            "lr copy": "LR Copy",
            "dc challan": "DC Challan",
            "weighment slip": "Weighment Slip",
            "e-way bill": "E-Way Bill",
            "eway bill": "E-Way Bill",
            "received acknowledgement form ed": "Received Acknowledgement Form ED",
            "received acknowledgement from ed": "Received Acknowledgement Form ED",
            "vehicle and material photo": "Vehicle and Material Photo",
            "vehicle and material": "Vehicle and Material Photo",

            # extra original columns
            "address line 1": "Address Line 1",
            "address line 2": "Address Line 2",
            "pincode": "Pincode",
            "city": "City",
            "state": "State",
            "country": "Country",
            "type": "Type",
            "total accepted qty in kg": "Total Accepted Qty",
            "value of accepted material in rs": "Value of Accepted Material (Old)",
            "rejection %": "Rejection %",
            "total dispatched qty in kg": "Total Dispatched Qty",
            "total value of dispatched materials in rs": "Total Value of Dispatched Materials",
            "total value of accepted qty in rs": "Total Value of Accepted Qty",
            "total incentive cost in rs": "Total Incentive Cost (Old)",
            "description": "Description",
            "wma": "WMA",
            "dc number": "DC Number",
            "invoice number": "Invoice Number",
            "s no": "S.no",
            "sno": "S.no",
            "s.no": "S.no",
        },
    },
}


# ============================================================
# 2) HELPER FUNCTIONS
# ============================================================

def canonical_col(name: str) -> str:
    """
    Normalise a column name so that similar spellings map together.
    """
    if name is None:
        return ""
    s = str(name).strip().lower()
    s = s.replace("_", " ").replace("-", " ")
    s = s.replace("/", " ")
    s = s.replace(".", " ")
    s = s.replace("(", " ").replace(")", " ")
    s = re.sub(r"[^a-z0-9\s]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s


def build_alias_lookup(dataset_name: str):
    aliases = CONFIG[dataset_name]["aliases"]
    lookup = {}
    for raw_key, target in aliases.items():
        lookup[canonical_col(raw_key)] = target
    return lookup


def detect_dataset(filename: str, columns):
    fname = filename.lower()

    for dataset_name, cfg in CONFIG.items():
        if any(k in fname for k in cfg["detect_keywords"]):
            return dataset_name

    colset = set(columns)
    for dataset_name, cfg in CONFIG.items():
        hits = sum(1 for c in cfg["detect_columns"] if c in colset)
        if hits >= 1:
            return dataset_name

    canon_cols = set(canonical_col(c) for c in columns)
    for dataset_name, cfg in CONFIG.items():
        detect_canon = set(canonical_col(c) for c in cfg["detect_columns"])
        hits = len(canon_cols.intersection(detect_canon))
        if hits >= 1:
            return dataset_name

    return None


def standardise_columns(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    lookup = build_alias_lookup(dataset_name)

    new_cols = []
    for col in df.columns:
        canon = canonical_col(col)
        std = lookup.get(canon, col)
        new_cols.append(std)

    df = df.copy()
    df.columns = new_cols
    return df


def coalesce_duplicate_columns(df: pd.DataFrame) -> pd.DataFrame:
    if df.columns.is_unique:
        return df

    out = pd.DataFrame(index=df.index)
    for col in df.columns.unique():
        same = df.loc[:, df.columns == col]

        if same.shape[1] == 1:
            out[col] = same.iloc[:, 0]
        else:
            combined = same.iloc[:, 0].copy()
            for j in range(1, same.shape[1]):
                combined = combined.combine_first(same.iloc[:, j])
            out[col] = combined

    return out


def apply_forward_fill(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    df = df.copy()
    non_ffill = set(CONFIG[dataset_name]["non_ffill_columns"])
    fill_cols = [c for c in df.columns if c not in non_ffill]

    if fill_cols:
        df[fill_cols] = df[fill_cols].ffill()

    return df


def to_numeric_safe(series):
    return pd.to_numeric(series, errors="coerce").fillna(0)


def split_cost_equally_by_code(df: pd.DataFrame, code_col: str, cost_cols: list) -> pd.DataFrame:
    """
    Split the given cost columns equally across all rows belonging to the same code.
    This OVERWRITES the original cost columns with split values.
    """
    df = df.copy()

    if code_col not in df.columns:
        return df

    # Make sure code column is usable
    code_series = df[code_col].astype(str).str.strip()

    # Count rows per code
    row_count = code_series.groupby(code_series).transform("size").replace(0, 1)

    for col in cost_cols:
        if col not in df.columns:
            df[col] = 0

        numeric_col = to_numeric_safe(df[col])

        # Take one original code-level cost (first non-null / first row after ffill)
        code_level_cost = numeric_col.groupby(code_series).transform("first")

        # Overwrite same column with equally split cost
        df[col] = (code_level_cost / row_count).round(2)

    return df


def add_derived_columns(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    """
    Add calculated business columns for Inward and Outward.
    Shared costs are split equally across rows of the same code.

    INWARD FORMULA:
    ---------------
    Net Procurement Cost = Value of Accepted Material + Transportation Cost
                           + Additional Cost + Loading Cost

    Logic: Net Procurement Cost is the TOTAL MONEY SPENT to procure material.
    It includes both the material purchase value AND all costs borne by the facility
    (transport, loading, additional). Always positive — higher means more was spent.

    Example:
      Material value = ₹1,000, Transport = ₹300 → Net Procurement Cost = ₹1,300
      Material value = ₹0 (non-valuable), Transport = ₹500 → Net Procurement Cost = ₹500
      (You spent ₹500 to collect waste with no material value)

    OUTWARD FORMULA:
    ----------------
    Net Material Sales Cost = Value of Accepted Material + Total Incentive Cost
                              - Transportation Cost - Additional Transport Cost
                              - Loading Cost

    Logic: Net Material Sales Cost is what the facility ACTUALLY EARNS after
    deducting all outward costs from the sale value. Can go negative if transport
    costs exceed sale value — meaning that shipment cost more than it earned.

    Example:
      Sale value = ₹1,000, Incentive = ₹100, Transport = ₹300
      → Net Sales = ₹1,000 + ₹100 - ₹300 = ₹800 (earned ₹800 net)

      Sale value = ₹200, Transport = ₹500
      → Net Sales = ₹200 - ₹500 = -₹300 (lost ₹300 on this shipment)
    """
    df = df.copy()

    if dataset_name == "Inward":
        for col in [
            "Received Quantity",
            "Rate",
            "Rejected Quantity",
            "Transportation Cost",
            "Additional Cost",
            "Loading Cost",
            "Inward Code",
        ]:
            if col not in df.columns:
                df[col] = 0

        # Split cost columns equally by Inward Code
        df = split_cost_equally_by_code(
            df,
            code_col="Inward Code",
            cost_cols=["Transportation Cost", "Additional Cost", "Loading Cost"]
        )

        received_qty = to_numeric_safe(df["Received Quantity"])
        rate = to_numeric_safe(df["Rate"])
        rejected_qty = to_numeric_safe(df["Rejected Quantity"])
        transportation_cost = to_numeric_safe(df["Transportation Cost"])
        additional_cost = to_numeric_safe(df["Additional Cost"])
        loading_cost = to_numeric_safe(df["Loading Cost"])

        df["Accepted Quantity"] = (received_qty - rejected_qty).clip(lower=0)
        df["Value of Received Material"] = (received_qty * rate).round(2)
        df["Value of Accepted Material"] = (df["Accepted Quantity"] * rate).round(2)

        # UPDATED FORMULA: Total cost to procure = material value + all costs borne
        # Always positive — represents total money spent to bring material in
        df["Net Procurement Cost"] = (
            df["Value of Accepted Material"]
            + transportation_cost
            + additional_cost
            + loading_cost
        ).round(2)

    elif dataset_name == "Outward":
        for col in [
            "Dispatched Quantity",
            "Rate",
            "Rejected Quantity",
            "Transportation Cost",
            "Additional Transport Cost",
            "Loading Cost",
            "Outward Code",
            "Incentive Cost per Kg",
        ]:
            if col not in df.columns:
                df[col] = 0

        # Split cost columns equally by Outward Code
        df = split_cost_equally_by_code(
            df,
            code_col="Outward Code",
            cost_cols=["Transportation Cost", "Additional Transport Cost", "Loading Cost"]
        )

        dispatched_qty = to_numeric_safe(df["Dispatched Quantity"])
        rate = to_numeric_safe(df["Rate"])
        rejected_qty = to_numeric_safe(df["Rejected Quantity"])
        transportation_cost = to_numeric_safe(df["Transportation Cost"])
        additional_transport_cost = to_numeric_safe(df["Additional Transport Cost"])
        loading_cost = to_numeric_safe(df["Loading Cost"])
        incentive_per_kg = to_numeric_safe(df["Incentive Cost per Kg"])

        df["Accepted Quantity"] = (dispatched_qty - rejected_qty).clip(lower=0)
        df["Value of Dispatched Material"] = (dispatched_qty * rate).round(2)
        df["Value of Accepted Material"] = (df["Accepted Quantity"] * rate).round(2)
        df["Total Incentive Cost"] = (incentive_per_kg * df["Accepted Quantity"]).round(2)

        # UNCHANGED FORMULA: Net earnings after deducting all outward costs
        # Can go negative if transport cost > sale value
        df["Net Material Sales Cost"] = (
            df["Value of Accepted Material"]
            + df["Total Incentive Cost"]
            - transportation_cost
            - additional_transport_cost
            - loading_cost
        ).round(2)

    return df


def keep_only_required_columns(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    df = df.copy()
    keep_cols = CONFIG[dataset_name]["keep_columns"]

    for col in keep_cols:
        if col not in df.columns:
            df[col] = pd.NA

    df = df[keep_cols]
    return df


def clean_dataset(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    """
    Full cleaning pipeline for one dataset.
    """
    df = standardise_columns(df, dataset_name)
    df = coalesce_duplicate_columns(df)
    df = apply_forward_fill(df, dataset_name)
    df = add_derived_columns(df, dataset_name)
    df = keep_only_required_columns(df, dataset_name)
    df = df.loc[:, ~df.columns.duplicated()]
    return df


# ============================================================
# 3) UPLOAD + PROCESS
# ============================================================

print("📂 Upload one or more CSV files (Inward / Production / Outward)...")
uploaded = files.upload()

cleaned_files = []

for filename, content in uploaded.items():
    try:
        print(f"\n🔍 Processing: {filename}")

        df = pd.read_csv(io.BytesIO(content))

        dataset_name = detect_dataset(filename, df.columns)
        if not dataset_name:
            print(f"⚠️ Could not detect dataset type for: {filename}")
            print("   Skipping this file.")
            continue

        print(f"✅ Detected dataset: {dataset_name}")
        print(f"   Original shape: {df.shape[0]} rows × {df.shape[1]} columns")

        cleaned_df = clean_dataset(df, dataset_name)

        print(f"   Cleaned shape:  {cleaned_df.shape[0]} rows × {cleaned_df.shape[1]} columns")
        print(f"   Final columns: {list(cleaned_df.columns)}")

        print("\n🧪 Sample cleaned output:")
        display(cleaned_df.head(10))

        output_filename = os.path.splitext(filename)[0] + "_cleaned.csv"
        cleaned_df.to_csv(output_filename, index=False)
        cleaned_files.append(output_filename)

        print(f"💾 Saved: {output_filename}")

    except Exception as e:
        print(f"❌ Error processing {filename}: {e}")


# ============================================================
# 4) ZIP ALL CLEANED FILES + DOWNLOAD
# ============================================================

if cleaned_files:
    zip_name = "cleaned_datasets.zip"
    with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as zf:
        for f in cleaned_files:
            zf.write(f)

    print(f"\n📦 Created ZIP: {zip_name}")
    files.download(zip_name)

else:
    print("\n⚠️ No cleaned files were generated.")
```

```
# -*- coding: utf-8 -*-
"""
wisewaste_data_engine.py
=========================
Merged replacement for the two separate Colab tools:
  - WiseWaste-Data-Standardization-Engine  (column cleaning, forward-fill,
    derived cost columns for Inward/Production/Outward)
  - enrich_facility_data.py  (Source Type / Vendor Type / Authorisation /
    Material Category enrichment for Inward/Outward)

One pass now does both, for whatever you actually upload:
  - A single CSV (e.g. just Inward.csv) works on its own — no other
    datasets required. You get back exactly that one cleaned/enriched file.
  - Any combination of standalone CSVs in one batch.
  - One combined multi-sheet Excel workbook (any subset of Inward /
    Production / Outward / Expenses / Revenue / Training present).

Removed vs. the original enrich_facility_data.py: the fragile exec()-based
attempt to pull mapping tables out of the DSE script at runtime. It never
actually worked (the DSE script never defined those tables), so
enrich_facility_data.py was always silently falling back to its own inline
copies anyway. Now there's exactly one copy of each table, defined once.

Usage (Colab):
    python wisewaste_data_engine.py   # then upload when prompted

Usage (local):
    python wisewaste_data_engine.py Inward.csv
    python wisewaste_data_engine.py Inward.csv Outward.csv Production.csv
    python wisewaste_data_engine.py Facility_Raw_Data.xlsx
"""
from __future__ import annotations

import io, os, re, sys, zipfile
import numpy as np
import pandas as pd

try:
    from google.colab import files
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# ================================================================
# 1) CONFIGURATION — column aliases, keep-lists, forward-fill rules
#    (Inward / Production / Outward only — Expense/Revenue/Training
#    pass through untouched except for date standardization, matching
#    both original scripts' behavior)
# ================================================================

CONFIG = {
    "Inward": {
        "detect_keywords": ["inward"],
        "detect_columns": ["Inward Code", "Received material from", "Project name"],
        "keep_columns": [
            "Inward Code",
            "Date",
            "Received Material From",
            "Vendor Location",
            "Project Name",
            "Vehicle Number",
            "Loading Cost",
            "Driver Name",
            "Vehicle Vendor Name",
            "Transportation Cost",
            "Additional Cost",
            "Total Tax",
            "Material",
            "Received Quantity",
            "Rate",
            "Value of Received Material",
            "Rejected Quantity",
            "Accepted Quantity",
            "Value of Accepted Material",
            "Net Procurement Cost",
            "Payment Status",
            "Weighment Slip",
            "Vendor Invoice",
            "Vehicle GPS Photo",
            "Inward Record Photo",
            "Material GPS Photo",
        ],
        "non_ffill_columns": {
            "Material",
            "Received Quantity",
            "Rate",
            "Amount",
            "Rejected Quantity",
            "Rejected Quantity %",
            "Accepted Billed Qty",
            "Value of Accepted Qty",
        },
        "aliases": {
            "inward code": "Inward Code",
            "date": "Date",
            "received material from": "Received Material From",
            "vendor location": "Vendor Location",
            "source": "Source",
            "project name": "Project Name",
            "vehicle number": "Vehicle Number",
            "loading cost": "Loading Cost",
            "driver name": "Driver Name",
            "vehicle vendor name": "Vehicle Vendor Name",
            "transportation cost": "Transportation Cost",
            "transportation cost in rs": "Transportation Cost",
            "additional cost": "Additional Cost",
            "additional cost in rs": "Additional Cost",
            "total tax": "Total Tax",
            "total tax in": "Total Tax",
            "material": "Material",
            "received quantity": "Received Quantity",
            "received quantity in kg": "Received Quantity",
            "rate": "Rate",
            "rate in rs": "Rate",
            "amount": "Amount",
            "amount in rs": "Amount",
            "rejected qty": "Rejected Quantity",
            "rejected qty in kg": "Rejected Quantity",
            "rejected quantity": "Rejected Quantity",
            "payment status": "Payment Status",
            "weighment slip": "Weighment Slip",
            "vendor invoice": "Vendor Invoice",
            "vehicle gps photo": "Vehicle GPS Photo",
            "inward record photo": "Inward Record Photo",
            "inward record": "Inward Record Photo",
            "material gps photo": "Material GPS Photo",
            "material gps": "Material GPS Photo",

            # extra original columns
            "rejected qty in": "Rejected Quantity %",
            "accepted billed qty in kg": "Accepted Billed Qty",
            "value of accepted qty in rs": "Value of Accepted Qty",
            "total received qty in kg": "Total Received Qty",
            "total accepted qty in kg": "Total Accepted Qty",
            "total rejected qty in kg": "Total Rejected Qty",
            "total amount in rs": "Total Amount",
            "dc number": "DC Number",
            "invoice number": "Invoice Number",
            "s no": "S.no",
            "sno": "S.no",
            "s.no": "S.no",
        },
    },

    "Production": {
        "detect_keywords": ["production"],
        "detect_columns": ["Production Code", "Process-Equipment", "No. of Present Staffs"],
        "keep_columns": [
            "Production Code",
            "Date",
            "Shift",
            "Process-Equipment",
            "Time Taken in Hrs",
            "No. of Staff Present",
            "Material",
            "Quantity in Kg",
            "Total Rejection in Kg",
            "Materials",
            "Material From",
            "Material Quantity",
        ],
        "non_ffill_columns": {
            "Material",
            "Quantity in Kg",
            "Total Rejection in Kg",
            "Materials",
            "Material From",
            "Material Quantity",
        },
        "aliases": {
            "production code": "Production Code",
            "date": "Date",
            "shift": "Shift",
            "process equipment": "Process-Equipment",
            "process-equipment": "Process-Equipment",
            "time taken in hrs": "Time Taken in Hrs",
            "time taken in hrs.": "Time Taken in Hrs",
            "time taken in": "Time Taken in Hrs",
            "time taken in hrs ": "Time Taken in Hrs",
            "time taken in hrs)": "Time Taken in Hrs",
            "time taken (in hrs)": "Time Taken in Hrs",
            "no of present staffs": "No. of Staff Present",
            "no. of present staffs": "No. of Staff Present",
            "no. of staff present": "No. of Staff Present",
            "material": "Material",
            "quantity in kg": "Quantity in Kg",
            "quantity (in kg)": "Quantity in Kg",
            "total rejection in kg": "Total Rejection in Kg",
            "total rejection (in kg)": "Total Rejection in Kg",
            "materials": "Materials",
            "material from": "Material From",
            "material quantity": "Material Quantity",
            "material quantity in kg": "Material Quantity",
            "material quantity (in kg)": "Material Quantity",

            # extra original columns
            "total quantity in kg": "Total Quantity",
            "total quantity (in kg)": "Total Quantity",
            "description": "Description",
            "s no": "S.no",
            "sno": "S.no",
            "s.no": "S.no",
        },
    },

    "Outward": {
        "detect_keywords": ["outward"],
        "detect_columns": ["Outward Code", "Vehicle number", "Dispatched Qty (in KG)"],
        "keep_columns": [
            "Outward Code",
            "Date",
            "Project",
            "Customer",
            "Destination",
            "Vehicle Number",
            "Transport Vendor",
            "Driver Name",
            "Driver Contact No",
            "Loading Cost",
            "Additional Transport Cost",
            "Transportation Cost",
            "Material",
            "Dispatched Quantity",
            "Rate",
            "Value of Dispatched Material",
            "Rejected Quantity",
            "Accepted Quantity",
            "Value of Accepted Material",
            "Tax",
            "Payment Status",
            "Incentive Cost per Kg",
            "Total Incentive Cost",
            "Net Material Sales Cost",
            "GPS Photo 1",
            "GPS Photo 2",
            "Invoice",
            "LR Copy",
            "DC Challan",
            "Weighment Slip",
            "E-Way Bill",
            "Received Acknowledgement Form ED",
            "Vehicle and Material Photo",
        ],
        "non_ffill_columns": {
            "Material",
            "Dispatched Quantity",
            "Rate",
            "Billed Amount",
            "Rejected Quantity",
            "Tax",
            "Incentive Cost per Kg",
        },
        "aliases": {
            "outward code": "Outward Code",
            "date": "Date",
            "project": "Project",
            "customer": "Customer",
            "destination": "Destination",
            "vehicle number": "Vehicle Number",
            "vehicle no": "Vehicle Number",
            "transport vendor": "Transport Vendor",
            "driver name": "Driver Name",
            "driver contact no": "Driver Contact No",
            "loading cost": "Loading Cost",
            "additional transport cost": "Additional Transport Cost",
            "transportation cost": "Transportation Cost",
            "transport cost": "Transportation Cost",
            "transport cost in rs": "Transportation Cost",
            "material": "Material",
            "dispatched quantity": "Dispatched Quantity",
            "dispatched qty in kg": "Dispatched Quantity",
            "dispatched qty": "Dispatched Quantity",
            "rate": "Rate",
            "rate in rs": "Rate",
            "billed amount": "Billed Amount",
            "billed amount in rs": "Billed Amount",
            "rejected quantity": "Rejected Quantity",
            "rejection qty in kg": "Rejected Quantity",
            "rejection qty": "Rejected Quantity",
            "tax": "Tax",
            "total tax in": "Tax",
            "payment status": "Payment Status",
            "incentive cost per kg": "Incentive Cost per Kg",
            "incentive costkg": "Incentive Cost per Kg",
            "gps photo 1": "GPS Photo 1",
            "gsp photo 1": "GPS Photo 1",
            "gps photo 2": "GPS Photo 2",
            "gsp photo 2": "GPS Photo 2",
            "invoice": "Invoice",
            "lr copy": "LR Copy",
            "dc challan": "DC Challan",
            "weighment slip": "Weighment Slip",
            "e-way bill": "E-Way Bill",
            "eway bill": "E-Way Bill",
            "received acknowledgement form ed": "Received Acknowledgement Form ED",
            "received acknowledgement from ed": "Received Acknowledgement Form ED",
            "vehicle and material photo": "Vehicle and Material Photo",
            "vehicle and material": "Vehicle and Material Photo",

            # extra original columns
            "address line 1": "Address Line 1",
            "address line 2": "Address Line 2",
            "pincode": "Pincode",
            "city": "City",
            "state": "State",
            "country": "Country",
            "type": "Type",
            "total accepted qty in kg": "Total Accepted Qty",
            "value of accepted material in rs": "Value of Accepted Material (Old)",
            "rejection %": "Rejection %",
            "total dispatched qty in kg": "Total Dispatched Qty",
            "total value of dispatched materials in rs": "Total Value of Dispatched Materials",
            "total value of accepted qty in rs": "Total Value of Accepted Qty",
            "total incentive cost in rs": "Total Incentive Cost (Old)",
            "description": "Description",
            "wma": "WMA",
            "dc number": "DC Number",
            "invoice number": "Invoice Number",
            "s no": "S.no",
            "sno": "S.no",
            "s.no": "S.no",
        },
    },
}

# ================================================================
# 2) MAPPING TABLES — source of truth for enrichment, extracted
#    verbatim from the original enrich_facility_data.py (update these
#    when Mapping_Sheet.xlsx changes, same as before)
# ================================================================

MATERIAL_CATEGORY_MAP = {
    "Unsorted Drywaste": "Unsorted Drywaste",
    "Coconut Shell": "Others",
    "Textile Waste": "Others",
    "Footwear": "Others",
    "Metal_Iron": "Others",
    "E-waste": "E-waste",
    "Wood": "Others",
    "Sanitary Waste": "Others",
    "Mattress": "Others",
    "Food Waste": "Rejects",
    "Rejects": "Rejects",
    "Water Rejects-PET": "Rejects",
    "Metal_Steel": "Metals",
    "Metal_Aluminum_Foil": "Metals",
    "Metal_Aluminum_Can": "Metals",
    "LVP (Co-Processing)": "Co-Processing",
    "SCF": "Co-Processing",
    "Mixed Flexible Plastics": "Flexible Plastics",
    "PP_Woven Bags": "Flexible Plastics",
    "Plastics_PP_Flexibles_Natural": "Flexible Plastics",
    "Plastics_PE_HM": "Flexible Plastics",
    "Plastics_LDPE_Grade 2": "Flexible Plastics",
    "Plastics_LDPE_Grade 1": "Flexible Plastics",
    "Plastics_PE_Covers_Oil Pouches": "Flexible Plastics",
    "Plastics_PE_Covers_Milk Pouches": "Flexible Plastics",
    "Plastics_PE_Covers_Milk Pouches_Akshayakalpa": "Flexible Plastics",
    "Plastics_PE_Covers_Milk Pouches_Heritage": "Flexible Plastics",
    "Plastics_MLP": "Flexible Plastics",
    "Plastics_MLP_Metalised": "Flexible Plastics",
    "Plastics_MLP_Non Metalised": "Flexible Plastics",
    "Plastics_Bale Straps": "Flexible Plastics",
    "Paper_Tissue": "Paper",
    "Paper_Paper Cup": "Paper",
    "Paper_Color Paper": "Paper",
    "Paper_White Paper": "Paper",
    "Paper_Tissue Roll": "Paper",
    "Paper_Carton Box": "Paper",
    "Plastics_PS_Foam": "PS",
    "Plastics_PS_Thermocol": "PS",
    "Mixed Rigid Plastics": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_Mixed": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_Mixed": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_Black": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_Black": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_White": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_White": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_Yellow": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_Yellow": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_Green": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_Green": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_Blue": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_Blue": "Rigid Plastics",
    "Plastics_HDPE_Grade 1_Red": "Rigid Plastics",
    "Plastics_HDPE_Grade 2_Red": "Rigid Plastics",
    "Plastics_HDPE_Brush": "Rigid Plastics",
    "Plastics_PP_Yellow": "Rigid Plastics",
    "Plastics_PP_Blue": "Rigid Plastics",
    "Plastics_PP_Green": "Rigid Plastics",
    "Plastics_PP_Red": "Rigid Plastics",
    "Plastics_PP_Black": "Rigid Plastics",
    "Plastics_PP_White": "Rigid Plastics",
    "Plastics_PP_Natural": "Rigid Plastics",
    "Plastics_PP_Mixed": "Rigid Plastics",
    "Plastics_PVC_Others_Natural": "Rigid Plastics",
    "Plastics_PVC_Pipe_Blue": "Rigid Plastics",
    "Plastics_PVC_Pipe_White": "Rigid Plastics",
    "Plastics_PVC_Pipe_Black": "Rigid Plastics",
    "Plastics_PETE_Others": "Rigid Plastics",
    "Plastics_PETE_Bottle_White": "Rigid Plastics",
    "Plastics_PETE_Bottle_Brown": "Rigid Plastics",
    "Plastics_PETE_Bottle_Green": "Rigid Plastics",
    "Plastics_PETE_Bottle_Grade 1_Natural": "Rigid Plastics",
    "Plastics_PETE_Bottle_Grade 2_Natural": "Rigid Plastics",
    "Plastics_PETE_Bottle_Grade 3_Natural": "Rigid Plastics",
    "Plastics_PETE Can": "Rigid Plastics",
    "Plastics_PETE Mixed": "Rigid Plastics",
    "Plastics_PPCP Buckets": "Rigid Plastics",
    "Plastics_LDPE_Mixed": "Rigid Plastics",
    "Plastics_LVP Rigid Plastic": "Rigid Plastics",
    "Plastics_ABS": "Rigid Plastics",
    "Plastics_PC": "Rigid Plastics",
    "Glass_Glass Bottle": "Glass",
    "Plastics_TetraPak": "TetraPak",
}

VENDOR_SOURCE_MAP = {
    ("Ajay Kumar - Gottigere", "Gottigere"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Anekal CC", "Anekal CC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Anekal CMC", "Anekal CMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Anekal Waste Pickers Colony", "Anekal"): {"source_type": "Waste Picker", "authorization": "Unregistered"},
    ("Asrafula Thubarahalli Munnekolala (WPC)", "Varthur"): {"source_type": "Waste Picker", "authorization": "Unregistered"},
    ("Attibele CC", "Attibele CC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Attibele DWCC", "Attibele DWCC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Attibele TMC", "Attibele TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 1 and 2"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 14,15"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 16,18 and 17"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 19,10,11,20 and 21"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 22,23 and 9"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 4,3,12 and 13"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 6"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Attibele Ward 7 and 8"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Commercial Route 2"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Commerical Route 1"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Attibele TMC", "Ward 5 and 6"): {"source_type": "ULB (Urban Local Body)", "authorization": "Registered"},
    ("Bagalur SS", "Bagalur SS"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Bagaluru GP", "Bagaluru"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Bagaluru GP", "Near SSR Concrete Blocks"): {"source_type": "SHG (Self Help Group)", "authorization": "Unregistered"},
    ("Bettahalsur DWCC", "EcoHUB Ecogram MRF"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Bijjawara GP", "Bijjawara GP"): {"source_type": "SHG (Self Help Group)", "authorization": "Unregistered"},
    ("BOMMASANDAR DWCC", "BOMMSANDRA DWCC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Bommasandra CC", "Bommasandra CC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Bommasandra CC", "Bommasandra DWCC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Bommasandra Collection Center", "Bommasandra Collection Center"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Bommasandra TMC", "Bommasandra TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("BSSHO Kannamangala CC", "Kannamangala CC"): {"source_type": "SHG (Self Help Group)", "authorization": "Unregistered"},
    ("Bukkasagara john", "Bukkasagara"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Chandapur TMC", "Hosur Road"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Chikkajala GP", "Chikkajala GP"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("CIPET", "Kudlu Gate"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("County", "Imperial Address"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("County", "Ittagur Village"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("Doddaballapur CMC", "Doddaballapur CMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Doddathogur GP", "Doddathogur"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Green Planet Recycling", "Green Planet Recycling (Begur)"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("HASSAN TRADERS", "HASSAN TRADERS"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Hebbagodi CC", "Hebbagodi CC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Hebbagodi CMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Hebbagodi TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 1"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 10"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 11"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 12"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 13"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 14"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 15"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 16"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 17"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 18"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 19"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 2"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 20"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 21"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 22"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 23"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 24"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 25"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 26"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 27"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 28"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 29"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 3"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 30"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 31"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 4"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 5"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 6"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 7"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 8"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi CMC", "Ward 9"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Hebbagodi DWCC", "Hebbagodi DWCC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Heebagodi cc", "Dwcc heebagodi"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Hurulichikkanahalli", "Hurulichikkanahalli"): {"source_type": "SHG (Self Help Group)", "authorization": "Unregistered"},
    ("Imperial Address", "Imperial Address"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("Imperial Address", "MC Halli, matnahalli, Panditana Agrahara"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("INCRESOL ENGINEERING SOLUTION PVt Ltd", "INCRESOL"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("India Hikes Pvt. Ltd.", "India Hikes Pvt. Ltd."): {"source_type": "Bulk Waste Generator", "authorization": "Registered"},
    ("India Hikes Pvt. Ltd", "India Hikes Pvt. Ltd"): {"source_type": "Bulk Waste Generator", "authorization": "Registered"},
    ("Insafe Industries", "Sy no 179/1 Singena Agrahara Village"): {"source_type": "Aggregators", "authorization": "Registered"},
    ("Islam Hulimangala Anekal (WPC)", "Hulimangala"): {"source_type": "Waste Picker", "authorization": "Unregistered"},
    ("Jadigenahalli GP", "Jadigenahalli GP"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Jamsticks India Pvt. Ltd.", "Jamsticks India Pvt. Ltd."): {"source_type": "Bulk Waste Generator", "authorization": "Registered"},
    ("Jigani CMC", "Jigani CMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Jigani DWCC", "Jigani DWCC"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Jigani TMC", "JIgani TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("JP Nagar", "JP Nagar"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Kalapsuddi Foundation", "Handenahalli"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("Kannamangala GP", "Kannamangala GP"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Kinelectro Lines India Private Limited", "Bommasandra, KIADB"): {"source_type": "Bulk Waste Generator", "authorization": "Registered"},
    ("Konappan Agrahar TMC", "Konappan Agrahar TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Kondarajanahalli GP", "Kondarajanahalli"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Konnappan Agrahar TMC", "Konnappan Agrahar TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("LAKSHMI ELECTRICALS", "LAKSHMI ELECTRICALS"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Madanayakana halli CMC", "Madanayakana halli CMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Madanayakanahalli CMC", "Madanayakanahalli TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Madani Traders", "Madani Traders"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Maranahalli SS", "Maranahalli SS"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Maranahalli TMC", "Maranahalli TMC"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marasur SS", "Marasur SS"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marenahalli GP", "Old Tavern Road"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Adesonnatti, Jai bhimnagar, Ishwarya, vakeel, and Advani layout"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Haleyuru, Shetty halli, Madivala"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Marsur 4th ward, JR Coconest"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Marsur GP"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Marsuru 3rd ward, Naganayakanahalli, Chandrakantha and MGB layout."): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Marsuru Agrahara, Bandapura, P C paradise, Trove layout, ZP layout"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur GP", "Marsuru ward 1 and ward 2"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Marsur SS", "Marsur SS"): {"source_type": "DWCC (Dry Waste Collection Center)", "authorization": "Unregistered"},
    ("Mayasandra GP", "Kamblipura, M.Medihalli & Mutsandra"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mayasandra GP", "Manchenahalli"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mayasandra GP", "Mayasandra GP"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mayasandra GP", "Ward 1, 3, 4 & Mayasandra)"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mayasandra GP", "Ward 2 & Bhaktipura"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mayasandra GP", "Ward 2, Bhakthipura, Medahalli, kambalipura, Muthsandra"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mayasandra SS", "Mayasndra ss"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mijanur Traders (WPC)", "Vinayak Layout"): {"source_type": "Waste Picker", "authorization": "Unregistered"},
    ("Mugaluru SS", "Mugaluru SS"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mugulur GP", "Kamanahalli, Chickanahalli, Rajivgandhi nagara"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mugulur GP", "Kuthganahalli, Bovipalya, Doddathimasandra"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mugulur GP", "MC Halli, matnahalli, Panditana Agrahara"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mugulur GP", "Mugulur, Guttahalli"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mugulur GP", "Shevaganaikanahalli, madapanaikanahalli, Banahalli, Kuguru"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Mugulur GP", "Thindlu"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("Paint Contractor Biyankah", "Jal vayu towers temple rd"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Paint Contractor Pratap Kumar", "Bengaluru"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Paint Contractor Rahul Singh", "Rahul Singh"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Painter Chaitra K", "Kammasandra"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Painter Bikas Gupta", "Choodasandra"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Painter Mahesh M", "Bommanahalli"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Painter Mrs Manisha", "Vajarahalli"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Painter Ramsurat Gupta", "Ilyas Nagar,Kumaraswamy Layout,"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Prasana Raju", "Near Freedom International School"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Prashanthan", "Bengaluru"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("PRF Jigani", "PRF Jigani/Attibele (Sarvam)"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("REBORNX Eco Tech Pvt Ltd", "Khata No 684/49,"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("RK Traders", "Housing Board Society Layout"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Sabina Traders", "Bettadasanpura"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("SahanaEnterprises", "106,8Th cross Road"): {"source_type": "Waste Picker", "authorization": "Unregistered"},
    ("Sholam Traders", "No 4,Begur Hobli"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Sugaturu GP", "Sugaturu"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
    ("SWMPL", "PRF Jigani"): {"source_type": "Aggregators", "authorization": "Registered"},
    ("SWMPL", "Sarvam MRF UNit 2 - Hargadde"): {"source_type": "Aggregators", "authorization": "Registered"},
    ("Tata Elxsi Limited", "Bhoruka Tech Park,"): {"source_type": "Bulk Waste Generator", "authorization": "Unregistered"},
    ("Textile Recovery Facility", "Textile Recovery Facility"): {"source_type": "Aggregators", "authorization": "Registered"},
    ("Trash to cash Innovations India Private Limited", "Sy No 64 2 Bassapura village"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("TRF Gottigere", "TRF Gottigere"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Venkatagiri kote GP", "Venkatagiri Kote"): {"source_type": "SHG (Self Help Group)", "authorization": "Unregistered"},
    ("ZWP-MRF Jigani", "No 4,Begur Hobli"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("ZWP-MRF Jigani", "ZWP-MRF-JIGANI"): {"source_type": "Aggregators", "authorization": "Unregistered"},
    ("Sahil Kumar Begur (WPC)", "Begur"): {"source_type": "Waste Picker", "authorization": "Unregistered"},
    ("Jalige Gp", "Jalige Gp"): {"source_type": "ULB (Urban Local Body)", "authorization": "Unregistered"},
}

CUSTOMER_DESTINATION_MAP = {
    ("Anekal TMC", "Anekal TMC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("Angirus Ind Pvt Ltd", "Angirus Ind Pvt Ltd"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("ARS Steels", "ARS Steel"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Attibele TMC", "Attibele TMC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("BANYAN SUSTAINABLE WASTE MANAGEMENT PRIVATE LIMITED", "BANYAN SUSTAINABLE WASTE MANAGEMENT PRIVATE LIMITED"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Barkathulla Traders", "Barkathulla Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Bhoomi Environmental Services", "Bhoomi Environmental Services"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Bommasandra CC", "Bommasandra CC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("Bommasandra TMC", "Bommasandra TMC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("CHAITHAN ENTERPRISES", "CHAITHAN ENTERPRISES"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Chandapura TMC", "Chandapura TMC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("CIPET (KCDC)", "CIPET (KCDC)"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("CLARITAS RECYCLERS PRIVATE LIMITED - UNIT - 2", "CLARITAS RECYCLERS PRIVATE LIMITED - UNIT - 2"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("Dalmia Cement", "Tadipatri Road,Jammaladugu Taluk"): {"facility_type": "Co-Processing", "authorization": "Registered"},
    ("DalmiaCement", "Tadipatri Road,Jammaladugu Taluk"): {"facility_type": "Co-Processing", "authorization": "Registered"},
    ("DELUXE RECYCLING", "DELEXUE RECYCLING"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Doddathogur GP", "Doddathogur GP"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("E Prakruti", "E Prakruti"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Eco India Recycling Solution", "Eco India Recycling Solution"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("ECOHANDLERS PRIVATE LIMITED", "Eco India Recycling Solution"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("ECOHANDLERS PRIVATE LIMITED", "ECOHANDLERS PRIVATE LIMITED (Sira)"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Fareed", "Fareed"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Gangadhara", "Gangadhara"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Green Planet Recycling", "GPR-Vittasandra (Old Facility)"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Green Planet Recycling", "Green Planet Recycling"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Hassan Traders", "Hassan Traders"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Hebbagodi CMC", "Hebbagodi CMC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("Isha Fiber and Fuel Source", "Isha Fiber and Fuel Source"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("JK CEMENT", "JK Cement - Bagalkot"): {"facility_type": "Co-Processing", "authorization": "Registered"},
    ("JR ECO TRADING PVT. LTD", "JR ECO TRADING PVT. LTD"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("JR ECO TRADING PVT. LTD", "MSGP"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Kalam Enterprise", "Kalam Enterprise"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("KGPCL", "KGPCL"): {"facility_type": "Co-Processing", "authorization": "Unregistered"},
    ("Krishna Polymach Pvt Limited", "Jigani main road"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("LAKSHMI ELECTRICALS", "LAKSHMI ELECTRICALS"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("LANDBELL GFS INDIA PRIVATE LIMITED", "LANDBELL GFS INDIA PRIVATE LIMITED"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("Lucky Enterprises", "1st Floor,No 1/296,14th street"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Madanayakanahalli", "Madanayakanhalli TMC"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("Madanayakanhalli", "Madanayakanhalli"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("Madani Traders", "Madani Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("MARASUR SS", "Marsuru SS"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("MarasuruGP", "JR Garden"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("MAYASANDRA GP", "Mayasandra GP"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("Mijanur Traders", "Mijanur Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("MM traders", "JK Cement - Bagalkot"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("MM traders", "MM traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("MSGP", "MSGP"): {"facility_type": "Co-Processing", "authorization": "Registered"},
    ("MSGP", "SN Traders - Salam"): {"facility_type": "Co-Processing", "authorization": "Registered"},
    ("Muguluru GP", "Telecom woods, phase 2, Doddathimmasandra"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("MUGULURU SS", "MUGULURU SS"): {"facility_type": "Returned to Source", "authorization": "Unregistered"},
    ("MysEchotchRecyclingPrivate limited", "32/1A1,Poonapali"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("NR ENTERPRISES", "NR ENTERPRISES"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Putul Traders", "Putul Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Recove Ventures Pvt Ltd", "Recove Ventures Pvt Ltd"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("REGENPLASTICS PRIVATE LIMITED", "REGENPLASTICS PRIVATE LIMITED"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Repon Traders", "Repon Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Rerout Private Limited", "Rerout Private Limited"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("RR Hydraulics", "RR Hydraulics"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("RUDRAYA WASTECH PRIVATE LIMITED", "RUDRAYA WASTECH PRIVATE LIMITED"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Sahana Enterprises", "Sahana Enterprises"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Sameer Traders", "Sameer Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Shadab Traders", "Shadab Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Shafiya Enterprises", "Honnakalasapura"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("SHAILA PLASTICS & KRISHI MITRA INDUSTRIES", "SHAILA PLASTICS & KRISHI MITRA INDUSTRIES"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("SN Traders (Salam)", "Repon Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("SN Traders (Salam)", "Shadab"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("SN Traders (Salam)", "SN Traders - Salam"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Sri Lakshmi Narashima Traders", "Sri Lakshmi Narashima Traders"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Srichakra Polyplast India pVT Ltd", "Sy No 355/A2,C,E"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("Srichakra Polyplast India Pvt Ltd", "Srichakra Polyplast India Pvt Ltd"): {"facility_type": "Recycler", "authorization": "Registered"},
    ("SWMPL", "Sarvam interm-PRF Jigani"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("SWMPL", "Sarvam-MRF Jigani"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("SWMPL", "TRF"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("SWMPL", "ZWP-MRF Jigani"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Tandav G Traders", "Tandav G Traders"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Trash To Cash Innovations India Private Limited", "Trash To Cash Innovations India Private Limited"): {"facility_type": "Aggregator", "authorization": "Unregistered"},
    ("Venkraft paper mills Pvt Ltd", "Venkraft paper mills Pvt Ltd-unit 2"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("Vinayak Industries", "Vinayak Industries"): {"facility_type": "Aggregator", "authorization": "Registered"},
    ("Vishuddh Recycle Private Limited", "Vishuddh Recycle Private Limited"): {"facility_type": "Recycler", "authorization": "Unregistered"},
    ("Krishna Polymach Pvt Ltd", "Krishna Polymach Pvt Ltd"): {"facility_type": "Recycler", "authorization": "Unregistered"},
}

# ================================================================
# 3) CLEANING PIPELINE (was: DSE)
# ================================================================

def canonical_col(name: str) -> str:
    """
    Normalise a column name so that similar spellings map together.
    """
    if name is None:
        return ""
    s = str(name).strip().lower()
    s = s.replace("_", " ").replace("-", " ")
    s = s.replace("/", " ")
    s = s.replace(".", " ")
    s = s.replace("(", " ").replace(")", " ")
    s = re.sub(r"[^a-z0-9\s]", " ", s)
    s = re.sub(r"\s+", " ", s).strip()
    return s


def build_alias_lookup(dataset_name: str):
    aliases = CONFIG[dataset_name]["aliases"]
    lookup = {}
    for raw_key, target in aliases.items():
        lookup[canonical_col(raw_key)] = target
    return lookup


def detect_dataset(filename: str, columns):
    fname = filename.lower()

    for dataset_name, cfg in CONFIG.items():
        if any(k in fname for k in cfg["detect_keywords"]):
            return dataset_name

    colset = set(columns)
    for dataset_name, cfg in CONFIG.items():
        hits = sum(1 for c in cfg["detect_columns"] if c in colset)
        if hits >= 1:
            return dataset_name

    canon_cols = set(canonical_col(c) for c in columns)
    for dataset_name, cfg in CONFIG.items():
        detect_canon = set(canonical_col(c) for c in cfg["detect_columns"])
        hits = len(canon_cols.intersection(detect_canon))
        if hits >= 1:
            return dataset_name

    return None


def standardise_columns(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    lookup = build_alias_lookup(dataset_name)

    new_cols = []
    for col in df.columns:
        canon = canonical_col(col)
        std = lookup.get(canon, col)
        new_cols.append(std)

    df = df.copy()
    df.columns = new_cols
    return df


def coalesce_duplicate_columns(df: pd.DataFrame) -> pd.DataFrame:
    if df.columns.is_unique:
        return df

    out = pd.DataFrame(index=df.index)
    for col in df.columns.unique():
        same = df.loc[:, df.columns == col]

        if same.shape[1] == 1:
            out[col] = same.iloc[:, 0]
        else:
            combined = same.iloc[:, 0].copy()
            for j in range(1, same.shape[1]):
                combined = combined.combine_first(same.iloc[:, j])
            out[col] = combined

    return out


def apply_forward_fill(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    df = df.copy()
    non_ffill = set(CONFIG[dataset_name]["non_ffill_columns"])
    fill_cols = [c for c in df.columns if c not in non_ffill]

    if fill_cols:
        df[fill_cols] = df[fill_cols].ffill()

    return df


def to_numeric_safe(series):
    return pd.to_numeric(series, errors="coerce").fillna(0)


def split_cost_equally_by_code(df: pd.DataFrame, code_col: str, cost_cols: list) -> pd.DataFrame:
    """
    Split the given cost columns equally across all rows belonging to the same code.
    This OVERWRITES the original cost columns with split values.
    """
    df = df.copy()

    if code_col not in df.columns:
        return df

    # Make sure code column is usable
    code_series = df[code_col].astype(str).str.strip()

    # Count rows per code
    row_count = code_series.groupby(code_series).transform("size").replace(0, 1)

    for col in cost_cols:
        if col not in df.columns:
            df[col] = 0

        numeric_col = to_numeric_safe(df[col])

        # Take one original code-level cost (first non-null / first row after ffill)
        code_level_cost = numeric_col.groupby(code_series).transform("first")

        # Overwrite same column with equally split cost
        df[col] = (code_level_cost / row_count).round(2)

    return df


def add_derived_columns(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    """
    Add calculated business columns for Inward and Outward.
    Shared costs are split equally across rows of the same code.

    INWARD FORMULA:
    ---------------
    Net Procurement Cost = Value of Accepted Material + Transportation Cost
                           + Additional Cost + Loading Cost

    Logic: Net Procurement Cost is the TOTAL MONEY SPENT to procure material.
    It includes both the material purchase value AND all costs borne by the facility
    (transport, loading, additional). Always positive — higher means more was spent.

    Example:
      Material value = ₹1,000, Transport = ₹300 → Net Procurement Cost = ₹1,300
      Material value = ₹0 (non-valuable), Transport = ₹500 → Net Procurement Cost = ₹500
      (You spent ₹500 to collect waste with no material value)

    OUTWARD FORMULA:
    ----------------
    Net Material Sales Cost = Value of Accepted Material + Total Incentive Cost
                              - Transportation Cost - Additional Transport Cost
                              - Loading Cost

    Logic: Net Material Sales Cost is what the facility ACTUALLY EARNS after
    deducting all outward costs from the sale value. Can go negative if transport
    costs exceed sale value — meaning that shipment cost more than it earned.

    Example:
      Sale value = ₹1,000, Incentive = ₹100, Transport = ₹300
      → Net Sales = ₹1,000 + ₹100 - ₹300 = ₹800 (earned ₹800 net)

      Sale value = ₹200, Transport = ₹500
      → Net Sales = ₹200 - ₹500 = -₹300 (lost ₹300 on this shipment)
    """
    df = df.copy()

    if dataset_name == "Inward":
        for col in [
            "Received Quantity",
            "Rate",
            "Rejected Quantity",
            "Transportation Cost",
            "Additional Cost",
            "Loading Cost",
            "Inward Code",
        ]:
            if col not in df.columns:
                df[col] = 0

        # Split cost columns equally by Inward Code
        df = split_cost_equally_by_code(
            df,
            code_col="Inward Code",
            cost_cols=["Transportation Cost", "Additional Cost", "Loading Cost"]
        )

        received_qty = to_numeric_safe(df["Received Quantity"])
        rate = to_numeric_safe(df["Rate"])
        rejected_qty = to_numeric_safe(df["Rejected Quantity"])
        transportation_cost = to_numeric_safe(df["Transportation Cost"])
        additional_cost = to_numeric_safe(df["Additional Cost"])
        loading_cost = to_numeric_safe(df["Loading Cost"])

        df["Accepted Quantity"] = (received_qty - rejected_qty).clip(lower=0)
        df["Value of Received Material"] = (received_qty * rate).round(2)
        df["Value of Accepted Material"] = (df["Accepted Quantity"] * rate).round(2)

        # UPDATED FORMULA: Total cost to procure = material value + all costs borne
        # Always positive — represents total money spent to bring material in
        df["Net Procurement Cost"] = (
            df["Value of Accepted Material"]
            + transportation_cost
            + additional_cost
            + loading_cost
        ).round(2)

    elif dataset_name == "Outward":
        for col in [
            "Dispatched Quantity",
            "Rate",
            "Rejected Quantity",
            "Transportation Cost",
            "Additional Transport Cost",
            "Loading Cost",
            "Outward Code",
            "Incentive Cost per Kg",
        ]:
            if col not in df.columns:
                df[col] = 0

        # Split cost columns equally by Outward Code
        df = split_cost_equally_by_code(
            df,
            code_col="Outward Code",
            cost_cols=["Transportation Cost", "Additional Transport Cost", "Loading Cost"]
        )

        dispatched_qty = to_numeric_safe(df["Dispatched Quantity"])
        rate = to_numeric_safe(df["Rate"])
        rejected_qty = to_numeric_safe(df["Rejected Quantity"])
        transportation_cost = to_numeric_safe(df["Transportation Cost"])
        additional_transport_cost = to_numeric_safe(df["Additional Transport Cost"])
        loading_cost = to_numeric_safe(df["Loading Cost"])
        incentive_per_kg = to_numeric_safe(df["Incentive Cost per Kg"])

        df["Accepted Quantity"] = (dispatched_qty - rejected_qty).clip(lower=0)
        df["Value of Dispatched Material"] = (dispatched_qty * rate).round(2)
        df["Value of Accepted Material"] = (df["Accepted Quantity"] * rate).round(2)
        df["Total Incentive Cost"] = (incentive_per_kg * df["Accepted Quantity"]).round(2)

        # UNCHANGED FORMULA: Net earnings after deducting all outward costs
        # Can go negative if transport cost > sale value
        df["Net Material Sales Cost"] = (
            df["Value of Accepted Material"]
            + df["Total Incentive Cost"]
            - transportation_cost
            - additional_transport_cost
            - loading_cost
        ).round(2)

    return df


def keep_only_required_columns(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    df = df.copy()
    keep_cols = CONFIG[dataset_name]["keep_columns"]

    for col in keep_cols:
        if col not in df.columns:
            df[col] = pd.NA

    df = df[keep_cols]
    return df


def clean_dataset(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    """
    Full cleaning pipeline for one dataset.
    """
    df = standardise_columns(df, dataset_name)
    df = coalesce_duplicate_columns(df)
    df = apply_forward_fill(df, dataset_name)
    df = add_derived_columns(df, dataset_name)
    df = keep_only_required_columns(df, dataset_name)
    df = df.loc[:, ~df.columns.duplicated()]
    return df




# ================================================================
# 4) ENRICHMENT (was: enrich_facility_data.py)
# ================================================================

def get_material_category(material_name):
    if not isinstance(material_name, str) or not material_name.strip():
        return "Unknown"
    m = material_name.strip()
    if m in MATERIAL_CATEGORY_MAP:
        return MATERIAL_CATEGORY_MAP[m]
    ml = m.lower()
    for k, v in MATERIAL_CATEGORY_MAP.items():
        if k.lower() == ml:
            return v
    for k, v in MATERIAL_CATEGORY_MAP.items():
        if k.lower() in ml or ml in k.lower():
            return v
    return "Unknown"


def get_source_type(vendor_name, vendor_location):
    if not isinstance(vendor_name, str) or not isinstance(vendor_location, str):
        return "Unknown"
    key = (vendor_name.strip(), vendor_location.strip())
    if key in VENDOR_SOURCE_MAP:
        return VENDOR_SOURCE_MAP[key]["source_type"]
    kl = (vendor_name.strip().lower(), vendor_location.strip().lower())
    for (k1, k2), v in VENDOR_SOURCE_MAP.items():
        if k1.lower() == kl[0] and k2.lower() == kl[1]:
            return v["source_type"]
    for (k1, k2), v in VENDOR_SOURCE_MAP.items():
        if k1.lower() == kl[0]:
            return v["source_type"]
    return "Unknown"


def get_vendor_authorization(vendor_name, vendor_location):
    if not isinstance(vendor_name, str) or not isinstance(vendor_location, str):
        return "Unknown"
    key = (vendor_name.strip(), vendor_location.strip())
    if key in VENDOR_SOURCE_MAP:
        return VENDOR_SOURCE_MAP[key]["authorization"]
    kl = (vendor_name.strip().lower(), vendor_location.strip().lower())
    for (k1, k2), v in VENDOR_SOURCE_MAP.items():
        if k1.lower() == kl[0] and k2.lower() == kl[1]:
            return v["authorization"]
    for (k1, k2), v in VENDOR_SOURCE_MAP.items():
        if k1.lower() == kl[0]:
            return v["authorization"]
    return "Unknown"


def get_vendor_type(customer_name, destination_name):
    if not isinstance(customer_name, str) or not isinstance(destination_name, str):
        return "Unknown"
    key = (customer_name.strip(), destination_name.strip())
    if key in CUSTOMER_DESTINATION_MAP:
        return CUSTOMER_DESTINATION_MAP[key]["facility_type"]
    kl = (customer_name.strip().lower(), destination_name.strip().lower())
    for (k1, k2), v in CUSTOMER_DESTINATION_MAP.items():
        if k1.lower() == kl[0] and k2.lower() == kl[1]:
            return v["facility_type"]
    for (k1, k2), v in CUSTOMER_DESTINATION_MAP.items():
        if k1.lower() == kl[0]:
            return v["facility_type"]
    return "Unknown"


def get_customer_authorization(customer_name, destination_name):
    if not isinstance(customer_name, str) or not isinstance(destination_name, str):
        return "Unknown"
    key = (customer_name.strip(), destination_name.strip())
    if key in CUSTOMER_DESTINATION_MAP:
        return CUSTOMER_DESTINATION_MAP[key]["authorization"]
    kl = (customer_name.strip().lower(), destination_name.strip().lower())
    for (k1, k2), v in CUSTOMER_DESTINATION_MAP.items():
        if k1.lower() == kl[0] and k2.lower() == kl[1]:
            return v["authorization"]
    for (k1, k2), v in CUSTOMER_DESTINATION_MAP.items():
        if k1.lower() == kl[0]:
            return v["authorization"]
    return "Unknown"


# ================================================================
# ENRICHMENT FUNCTIONS
# ================================================================

def standardize_dates(df):
    """Convert all date columns to DD-MM-YYYY format (no time component)."""
    df = df.copy()
    date_cols = [c for c in df.columns if 'date' in c.lower()]
    for col in date_cols:
        try:
            df[col] = pd.to_datetime(df[col], dayfirst=True, errors='coerce')
            df[col] = df[col].dt.strftime('%d-%m-%Y')
            # Replace NaT string with empty
            df[col] = df[col].replace('NaT', '')
        except Exception:
            pass
    return df


def enrich_inward(df):
    """Add Source Type, Authorisation, Inward Material Category to Inward sheet."""
    df = df.copy()

    # Find columns (flexible matching)
    vendor_col = next((c for c in df.columns if 'received material from' in c.lower()), None)
    location_col = next((c for c in df.columns if 'vendor location' in c.lower()), None)
    material_col = next((c for c in df.columns if c.lower() == 'material'), None)

    if vendor_col and location_col and 'Source Type' not in df.columns:
        loc_pos = list(df.columns).index(location_col) + 1
        df.insert(loc_pos, 'Source Type',
            df.apply(lambda r: get_source_type(
                str(r.get(vendor_col, '')), str(r.get(location_col, ''))), axis=1))
        df.insert(loc_pos + 1, 'Authorisation',
            df.apply(lambda r: get_vendor_authorization(
                str(r.get(vendor_col, '')), str(r.get(location_col, ''))), axis=1))
        print(f"  ✅ Added Source Type + Authorisation after '{location_col}'")
    else:
        print(f"  ⚠️  Could not find vendor/location columns")

    if material_col and 'Inward Material Category' not in df.columns:
        mat_pos = list(df.columns).index(material_col) + 1
        df.insert(mat_pos, 'Inward Material Category',
            df[material_col].apply(get_material_category))
        print(f"  ✅ Added Inward Material Category after '{material_col}'")

    return df


def enrich_outward(df):
    """Add Vendor Type, Authorisation, Outward Material Category to Outward sheet."""
    df = df.copy()

    customer_col = next((c for c in df.columns if c.lower() == 'customer'), None)
    dest_col = next((c for c in df.columns if c.lower() == 'destination'), None)
    material_col = next((c for c in df.columns if c.lower() == 'material'), None)

    if customer_col and dest_col:
        dest_pos = list(df.columns).index(dest_col) + 1
        df.insert(dest_pos, 'Vendor Type',
            df.apply(lambda r: get_vendor_type(
                str(r.get(customer_col, '')), str(r.get(dest_col, ''))), axis=1))
        df.insert(dest_pos + 1, 'Authorisation',
            df.apply(lambda r: get_customer_authorization(
                str(r.get(customer_col, '')), str(r.get(dest_col, ''))), axis=1))
        print(f"  ✅ Added Vendor Type + Authorisation after '{dest_col}'")
    else:
        print(f"  ⚠️  Could not find customer/destination columns")

    if material_col:
        mat_pos = list(df.columns).index(material_col) + 1
        df.insert(mat_pos, 'Outward Material Category',
            df[material_col].apply(get_material_category))
        print(f"  ✅ Added Outward Material Category after '{material_col}'")

    return df





# ================================================================
# UNIFIED ORCHESTRATION
# ================================================================
# Ties the two engines together:
#   1. Column standardization / forward-fill / derived costs (was: DSE)
#   2. Source Type / Vendor Type / Authorisation / Material Category
#      enrichment (was: enrich_facility_data)
# into one pass per dataset, and removes the fragile cross-script
# exec()-import that enrich_facility_data.py used to pull mapping tables
# out of the DSE script at runtime (it never actually worked — the DSE
# script never defined those tables — enrich_facility_data.py was
# silently falling back to its own inline copies every time anyway).
# Both engines' full inline mapping tables are used directly here, in
# one place, since the merge makes the fallback moot.
#
# Dataset types fall into two buckets:
#   - CLEANED + ENRICHED  : Inward, Outward  (get the full pipeline)
#   - CLEANED ONLY        : Production        (DSE cleaning, no vendor/
#                                               customer mapping applies)
#   - PASSTHROUGH         : Expenses, Revenue, Training, and anything
#                            unrecognized (only dates get standardized —
#                            matches both original scripts' behavior,
#                            since neither one applied column cleaning to
#                            these three)
#
# "No restrictions on partial uploads" means: every entry point here
# accepts whatever is present — one CSV, a handful of CSVs, or one
# multi-sheet workbook with any subset of the six sheets — and returns
# exactly what was given, cleaned/enriched where applicable. Nothing
# requires the full set of five datasets to be present.

CLEAN_AND_ENRICH = {"Inward", "Outward"}
CLEAN_ONLY = {"Production"}
PASSTHROUGH_TYPES = {"Expense", "Revenue", "Training"}

# DSE's own detect_dataset() only knows about Inward/Production/Outward
# (its CONFIG dict has no entries for the other three). These are the
# extra, lightweight detection rules layered on top for Expense/Revenue/
# Training, so a lone "Expense.csv" upload is recognized on its own
# rather than falling through to "unknown" and being skipped.
_PASSTHROUGH_DETECT = {
    "Expense":  {"keywords": ["expense"],  "columns": ["Expense Code"]},
    "Revenue":  {"keywords": ["revenue"],  "columns": ["Revenue Code"]},
    "Training": {"keywords": ["training"], "columns": ["Training Code"]},
}

# Sheet-name variants seen in real exports (e.g. "Expenses" vs "Expense")
# mapped to the canonical dataset name used throughout this script.
_SHEET_NAME_ALIASES = {
    "inward": "Inward", "production": "Production", "outward": "Outward",
    "expense": "Expense", "expenses": "Expense",
    "revenue": "Revenue", "training": "Training",
}


def detect_dataset_any(filename: str, columns) -> str | None:
    """Extends DSE's detect_dataset() to also recognize Expense/Revenue/
    Training — by filename keyword first, then by a signature column."""
    hit = detect_dataset(filename, columns)
    if hit:
        return hit
    fname = filename.lower()
    colset = set(columns)
    for name, rule in _PASSTHROUGH_DETECT.items():
        if any(k in fname for k in rule["keywords"]):
            return name
    for name, rule in _PASSTHROUGH_DETECT.items():
        if any(c in colset for c in rule["columns"]):
            return name
    return None


def process_dataset(df: pd.DataFrame, dataset_name: str) -> pd.DataFrame:
    """Runs the right pipeline for one dataset, by name. This is the single
    function both the CSV path and the workbook path funnel through, so
    there's exactly one place that decides what 'processing Inward' means."""
    if dataset_name in CLEAN_AND_ENRICH:
        df = clean_dataset(df, dataset_name)  # DSE: standardize/ffill/derive
        df = enrich_inward(df) if dataset_name == "Inward" else enrich_outward(df)
    elif dataset_name in CLEAN_ONLY:
        df = clean_dataset(df, dataset_name)
    # PASSTHROUGH_TYPES and anything unrecognized: no column changes.
    df = standardize_dates(df)
    return df


def _report_mapping_coverage(dataset_name: str, df: pd.DataFrame):
    col = "Source Type" if dataset_name == "Inward" else "Vendor Type" if dataset_name == "Outward" else None
    if not col or col not in df.columns or len(df) == 0:
        return
    unk = (df[col] == "Unknown").sum()
    pct = (len(df) - unk) / len(df) * 100
    print(f"  📋 {dataset_name} mapping coverage: {len(df)-unk}/{len(df)} ({pct:.1f}%)")
    if unk > 0:
        id_cols = (["Received Material From", "Vendor Location"] if dataset_name == "Inward"
                   else ["Customer", "Destination"])
        id_cols = [c for c in id_cols if c in df.columns]
        if id_cols:
            miss = df[df[col] == "Unknown"][id_cols].drop_duplicates()
            print(f"  ⚠️  {unk} rows with Unknown {col} — unmapped pairs:")
            print(miss.to_string(index=False))


def process_single_file(filename: str, file_bytes: bytes = None, file_path: str = None) -> tuple:
    """One standalone CSV (or single-sheet Excel) in, one cleaned/enriched
    dataframe out — no other datasets required. Returns (dataset_name, df),
    or (None, None) if the type couldn't be detected at all."""
    if file_bytes is not None:
        buf = io.BytesIO(file_bytes)
    else:
        buf = file_path

    if str(filename).lower().endswith((".xlsx", ".xls")):
        df = pd.read_excel(buf)
    else:
        df = pd.read_csv(buf)

    dataset_name = detect_dataset_any(filename, df.columns)
    if not dataset_name:
        print(f"⚠️  Could not detect dataset type for '{filename}' — skipping.")
        return None, None

    print(f"✅ Detected: {dataset_name}  ({df.shape[0]} rows × {df.shape[1]} cols)")
    df = process_dataset(df, dataset_name)
    print(f"   → cleaned:  {df.shape[0]} rows × {df.shape[1]} cols")
    _report_mapping_coverage(dataset_name, df)
    return dataset_name, df


def process_workbook(filename: str, file_bytes: bytes = None, file_path: str = None) -> dict:
    """One multi-sheet Excel in (any subset of Inward/Production/Outward/
    Expenses/Revenue/Training — missing sheets are simply not in the
    output, nothing requires all six), one {sheet_name: cleaned_df} dict
    out, using each sheet's ORIGINAL name as the output key so downstream
    consumers see the same sheet names they uploaded."""
    xl = pd.ExcelFile(io.BytesIO(file_bytes) if file_bytes is not None else file_path)
    output_sheets = {}

    for sheet in xl.sheet_names:
        df = pd.read_excel(xl, sheet_name=sheet)
        if df.empty:
            output_sheets[sheet] = df
            continue
        canonical = _SHEET_NAME_ALIASES.get(sheet.strip().lower())
        if canonical is None:
            # Unrecognized sheet name — try column-signature detection
            # before giving up and passing it through untouched.
            canonical = detect_dataset_any(sheet, df.columns)
        print(f"\n  Sheet: {sheet}" + (f"  (→ {canonical})" if canonical else "  (unrecognized — passthrough)"))
        print(f"  Input:  {len(df)} rows, {len(df.columns)} cols")
        df = process_dataset(df, canonical) if canonical else standardize_dates(df)
        print(f"  Output: {len(df)} rows, {len(df.columns)} cols")
        if canonical:
            _report_mapping_coverage(canonical, df)
        output_sheets[sheet] = df

    return output_sheets


def save_single_output(dataset_name: str, df: pd.DataFrame, base_filename: str, out_dir: str = ".") -> str:
    base = os.path.splitext(os.path.basename(base_filename))[0]
    out_path = os.path.join(out_dir, f"{base}_cleaned.csv")
    df.to_csv(out_path, index=False)
    print(f"💾 Saved: {out_path}")
    return out_path


def save_workbook_output(output_sheets: dict, base_filename: str, out_dir: str = ".") -> str:
    base = os.path.splitext(os.path.basename(base_filename))[0]
    out_path = os.path.join(out_dir, f"{base}_enriched.xlsx")
    with pd.ExcelWriter(out_path, engine="openpyxl") as writer:
        for sheet, df in output_sheets.items():
            df.to_excel(writer, sheet_name=sheet, index=False)
    print(f"\n✅ Saved: {out_path}")
    return out_path


# ================================================================
# ENTRY POINT — accepts any mix of standalone files and workbooks,
# no restriction on which datasets are present.
# ================================================================

def process_uploads(uploads: dict, out_dir: str = ".") -> list:
    """`uploads` is {filename: file_bytes}. Handles any combination:
    a single Inward.csv on its own, five separate CSVs, one combined
    workbook, or a mix of workbooks and standalone CSVs in the same
    batch. Returns the list of output file paths written."""
    written = []
    for filename, file_bytes in uploads.items():
        print(f"\n📊 Processing: {filename}")
        try:
            if filename.lower().endswith((".xlsx", ".xls")):
                xl = pd.ExcelFile(io.BytesIO(file_bytes))
                if len(xl.sheet_names) > 1:
                    output_sheets = process_workbook(filename, file_bytes=file_bytes)
                    written.append(save_workbook_output(output_sheets, filename, out_dir))
                    continue
            dataset_name, df = process_single_file(filename, file_bytes=file_bytes)
            if dataset_name is not None:
                written.append(save_single_output(dataset_name, df, filename, out_dir))
        except Exception as e:
            print(f"❌ Error processing {filename}: {e}")
    return written


if __name__ == "__main__":
    if IN_COLAB:
        print("📂 Upload any combination of files: standalone CSVs (Inward/Production/")
        print("   Outward/Expense/Revenue/Training) and/or a combined multi-sheet Excel.")
        print("   Upload just one file if that's all you have — no restrictions.")
        uploaded = files.upload()
        outputs = process_uploads(uploaded, out_dir="/tmp")
        if outputs:
            zip_name = "/tmp/cleaned_datasets.zip"
            with zipfile.ZipFile(zip_name, "w", zipfile.ZIP_DEFLATED) as zf:
                for f in outputs:
                    zf.write(f, arcname=os.path.basename(f))
            print(f"\n📦 Created ZIP: {zip_name}")
            files.download(zip_name)
        else:
            print("\n⚠️  No files were successfully processed.")
    else:
        if len(sys.argv) < 2:
            print("Usage: python wisewaste_data_engine.py <file1> [file2 ...]")
            print("Accepts any mix of standalone CSVs and multi-sheet Excel workbooks.")
            sys.exit(1)
        uploads = {}
        for path in sys.argv[1:]:
            with open(path, "rb") as f:
                uploads[os.path.basename(path)] = f.read()
        process_uploads(uploads, out_dir=".")
```
